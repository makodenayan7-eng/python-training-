```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

data={
    'name':['A','B','C','D','E'],
    'math':[78,67,96,36,65],
    'science':[78,67,96,36,65],
    'english':[78,67,96,36,65]
}
df=pd.DataFrame(data)
print(df)
```

```python
marks = df[["english", "math", "science"]]
Average_marks = np.mean(marks, axis=1)
df["average"] = Average_marks
print(df)
```

```python
def grade(avg):
  if avg>=90:
    return 'A'
  elif avg>=75:
    return 'B'
  elif avg>=60:
    return 'C'
  else:
    return 'D'

df['grade']= df["average"]
```

```python
plt.figure()
plt.bar(df["name"],df["average"])
plt.xlabel('students')
plt.ylabel('Average marks')
plt.title('Average marks of students')
plt.show()
```
