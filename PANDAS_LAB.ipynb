```python
import numpy as np
students =np.array([
    [85,90,78], # student1
    [60,55,70], # student2
    [95,92,88] ,# student3
    [40,50,45], # student4
    [72,68,75] # student5
])
students
```

```python
avg_marks=np.mean(students,axis=1)
print("Average Marks:",avg_marks)
```

```python
topper= np.argmax(avg_marks)
print("topper student:",topper)
```

```python
failed_student = np.where(avg_marks<50)
print("Failed Students:",failed_student[0]+1)
```

```python
#performance report
for i, avg in enumerate(avg_marks):
  status="pass"if avg>50 else"fail"
  print(f"student{i+1}:avg={avg:.2f},{status}")
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[2,4,6,8]
plt.plot(x,y)
```

```python
x=[1,2,3,4]
y=[2,1,4,1]
plt.plot(x,y)
plt.title('line plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.grid(True)
plt.show()
```

```python
#BAR CHART
months =['jan','feb','mar','apr','may']
sales=[100,150,75,200,120]
plt.bar(months,sales,color='black')
plt.title('monthly sales')
plt.xlabel('months')
plt.ylabel('sales')
plt.show()
```

```python
#SCATTER PLOT(COMPARING HEIGHTVS WEIGHT)
height=[150,160,170,180]
weight=[50,60,70,80]
plt.scatter(height,weight)
plt.title('height vs weight')
plt.xlabel('height')
plt.ylabel('weight')
plt.show()
```

```python
#pie chart(market share)
brands=['tata','mahindra','suzuki']
shares=[45,25,30]
plt.pie(shares,labels=brands,autopct='%1.1f%%')
plt.title('market share')
plt.show()
```

```python
#multiple plot in single figure
x=[1,2,3,4]
y1=[2,4,6,8]
y2=[3,6,9,12]
plt.plot(x,y1,label='square')
plt.plot(x,y2,label='linear')
plt.legend()
plt.title('multiple plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
#subplots
fig, axs = plt.subplots(2,2)

axs[0,0].plot(x,y1)
axs[0,0].set_title('square')
axs[0,1].bar(months,sales)
axs[0,1].set_title('sales')
axs[1,0].scatter(height,weight)
axs[1,0].set_title('scatter')
axs[1,1].pie(shares,labels=brands,autopct='%1.1f%%')
axs[1,1].set_title('pie')
plt.tight_layout()
plt.show()
#
```
