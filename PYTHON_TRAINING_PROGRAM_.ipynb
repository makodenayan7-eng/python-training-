```python
#Aceept 5 paper marks,roll no,name:
and calculate total& percentage
if percentage >=85,student can take admission
else can't take admission
using class & object &constructor
```

```python
class Student:
    def __init__(self, rollno, name):
        self.rollno = rollno
        self.name = name

    def accept(self):
        self.m1 = int(input("Enter paper 1 marks: "))
        self.m2 = int(input("Enter paper 2 marks: "))
        self.m3 = int(input("Enter paper 3 marks: "))
        self.m4 = int(input("Enter paper 4 marks: "))
        self.m5 = int(input("Enter paper 5 marks: "))

    def calculate(self):
        total = self.m1 + self.m2 + self.m3 + self.m4 + self.m5
        percentage = (total / 500) * 100

        print("\n----- Student Details -----")
        print("Roll Number:", self.rollno)
        print("Name:", self.name)
        print("Total Marks:", total)
        print("Percentage:", percentage, "%")

        if percentage >= 85:
            print("Student can take admission")
        else:
            print("Student can't take admission")



s1 = Student(51, "NAYAN")


s1.accept()
s1.calculate()
```
